```python
data.columns
```

```python
columns_to_keep=['name', 'online_order', 'book_table', 'rate','dish_liked','approx_cost(for two people)']
columns_to_drop=['url', 'address','votes','phone', 'location', 'rest_type','cuisines','reviews_list', 'menu_item',
       'listed_in(type)', 'listed_in(city)']
```

```python
len(columns_to_keep)
```

```python
len(columns_to_drop)
```

```python
#dala cleaning
```

```python
data.drop(columns=columns_to_drop,inplace=True)
```

```python
data
```

```python
#Renaming the columns
```

```python
data.rename(columns={"name":"customer name"})
```

```python
for i in data.columns:
    print(i.capitalize())
```

```python
neww_columns_names=[]
for i in data.columns:
    neww_columns_names.append(i.capitalize())
```

```python
neww_columns_names
```

```python
data.columns=neww_columns_names
```

```python
data
```

```python
#Dropping duplicates.
```

```python
data.duplicated().sum()
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.duplicated().sum()
```

```python
data.isna().sum()
```

```python
data.dropna(inplace=True)
```

```python
data.isna().sum()
```

```python
data
```

```python
data.reset_index(inplace=True)
```

```python
data.drop(columns="index", inplace=True)
```

```python
data["Rate"]=data["Rate"].str.replace(" ","")
```

```python
data["Rate"]=data["Rate"].str.replace("/"," out of ")
```

```python
data["Online_order"]
```

```python
data["Online_order"]=data["Online_order"].apply(lambda x:1 if x=="Yes" else 0)
```
